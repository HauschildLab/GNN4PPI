## Final tests for paper (copied from KIRC)

```python
combinations = list(product(np.arange(10), np.arange(5)))
df = pd.DataFrame(combinations, columns=['test_round', 'es_fold'])

for i in range(50):
    #full data
    df.loc[(df['test_round']== MLP_outer_ext.iloc[i]['config/train_loop_config/dataset_test_round']) & (df['es_fold']== MLP_outer_ext.iloc[i]['config/train_loop_config/es_folds']), 'MLP'] = MLP_outer_ext.iloc[i]['test_mcc']
    df.loc[(df['test_round']== GCN_outer_ext.iloc[i]['config/train_loop_config/dataset_test_round']) & (df['es_fold']== GCN_outer_ext.iloc[i]['config/train_loop_config/es_folds']), 'GCN'] = GCN_outer_ext.iloc[i]['test_mcc']
    df.loc[(df['test_round']== GAT_outer_ext.iloc[i]['config/train_loop_config/dataset_test_round']) & (df['es_fold']== GAT_outer_ext.iloc[i]['config/train_loop_config/es_folds']), 'GAT'] = GAT_outer_ext.iloc[i]['test_mcc']
    df.loc[(df['test_round']== ChebNet_outer_ext.iloc[i]['config/train_loop_config/dataset_test_round']) & (df['es_fold']== ChebNet_outer_ext.iloc[i]['config/train_loop_config/es_folds']), 'ChebNet'] = ChebNet_outer_ext.iloc[i]['test_mcc']
    df.loc[(df['test_round']== GCN2MLP_outer_ext.iloc[i]['config/train_loop_config/dataset_test_round']) & (df['es_fold']== GCN2MLP_outer_ext.iloc[i]['config/train_loop_config/es_folds']), 'GCN2MLP'] = GCN2MLP_outer_ext.iloc[i]['test_mcc']
    df.loc[(df['test_round']== GAT2MLP_outer_ext.iloc[i]['config/train_loop_config/dataset_test_round']) & (df['es_fold']== GAT2MLP_outer_ext.iloc[i]['config/train_loop_config/es_folds']), 'GAT2MLP'] = GAT2MLP_outer_ext.iloc[i]['test_mcc']
    df.loc[(df['test_round']== ChebNet2MLP_outer_ext.iloc[i]['config/train_loop_config/dataset_test_round']) & (df['es_fold']== ChebNet2MLP_outer_ext.iloc[i]['config/train_loop_config/es_folds']), 'ChebNet2MLP'] = ChebNet2MLP_outer_ext.iloc[i]['test_mcc']

    #methylation data
    df.loc[(df['test_round']== mMLP_outer_ext.iloc[i]['config/train_loop_config/dataset_test_round']) & (df['es_fold']== mMLP_outer_ext.iloc[i]['config/train_loop_config/es_folds']), 'mMLP'] = mMLP_outer_ext.iloc[i]['test_mcc']
    df.loc[(df['test_round']== mGCN_outer_ext.iloc[i]['config/train_loop_config/dataset_test_round']) & (df['es_fold']== mGCN_outer_ext.iloc[i]['config/train_loop_config/es_folds']), 'mGCN'] = mGCN_outer_ext.iloc[i]['test_mcc']
    df.loc[(df['test_round']== mGAT_outer_ext.iloc[i]['config/train_loop_config/dataset_test_round']) & (df['es_fold']== mGAT_outer_ext.iloc[i]['config/train_loop_config/es_folds']), 'mGAT'] = mGAT_outer_ext.iloc[i]['test_mcc']
    df.loc[(df['test_round']== mChebNet_outer_ext.iloc[i]['config/train_loop_config/dataset_test_round']) & (df['es_fold']== mChebNet_outer_ext.iloc[i]['config/train_loop_config/es_folds']), 'mChebNet'] = mChebNet_outer_ext.iloc[i]['test_mcc']
    df.loc[(df['test_round']== mGCN2MLP_outer_ext.iloc[i]['config/train_loop_config/dataset_test_round']) & (df['es_fold']== mGCN2MLP_outer_ext.iloc[i]['config/train_loop_config/es_folds']), 'mGCN2MLP'] = mGCN2MLP_outer_ext.iloc[i]['test_mcc']
    df.loc[(df['test_round']== mGAT2MLP_outer_ext.iloc[i]['config/train_loop_config/dataset_test_round']) & (df['es_fold']== mGAT2MLP_outer_ext.iloc[i]['config/train_loop_config/es_folds']), 'mGAT2MLP'] = mGAT2MLP_outer_ext.iloc[i]['test_mcc']
    df.loc[(df['test_round']== mChebNet2MLP_outer_ext.iloc[i]['config/train_loop_config/dataset_test_round']) & (df['es_fold']== mChebNet2MLP_outer_ext.iloc[i]['config/train_loop_config/es_folds']), 'mChebNet2MLP'] = mChebNet2MLP_outer_ext.iloc[i]['test_mcc']

    #expression data
    df.loc[(df['test_round']== xMLP_outer_ext.iloc[i]['config/train_loop_config/dataset_test_round']) & (df['es_fold']== xMLP_outer_ext.iloc[i]['config/train_loop_config/es_folds']), 'xMLP'] = xMLP_outer_ext.iloc[i]['test_mcc']
    df.loc[(df['test_round']== xGCN_outer_ext.iloc[i]['config/train_loop_config/dataset_test_round']) & (df['es_fold']== xGCN_outer_ext.iloc[i]['config/train_loop_config/es_folds']), 'xGCN'] = xGCN_outer_ext.iloc[i]['test_mcc']
    df.loc[(df['test_round']== xGAT_outer_ext.iloc[i]['config/train_loop_config/dataset_test_round']) & (df['es_fold']== xGAT_outer_ext.iloc[i]['config/train_loop_config/es_folds']), 'xGAT'] = xGAT_outer_ext.iloc[i]['test_mcc']
    df.loc[(df['test_round']== xChebNet_outer_ext.iloc[i]['config/train_loop_config/dataset_test_round']) & (df['es_fold']== xChebNet_outer_ext.iloc[i]['config/train_loop_config/es_folds']), 'xChebNet'] = xChebNet_outer_ext.iloc[i]['test_mcc']
    df.loc[(df['test_round']== xGCN2MLP_outer_ext.iloc[i]['config/train_loop_config/dataset_test_round']) & (df['es_fold']== xGCN2MLP_outer_ext.iloc[i]['config/train_loop_config/es_folds']), 'xGCN2MLP'] = xGCN2MLP_outer_ext.iloc[i]['test_mcc']
    df.loc[(df['test_round']== xGAT2MLP_outer_ext.iloc[i]['config/train_loop_config/dataset_test_round']) & (df['es_fold']== xGAT2MLP_outer_ext.iloc[i]['config/train_loop_config/es_folds']), 'xGAT2MLP'] = xGAT2MLP_outer_ext.iloc[i]['test_mcc']
    df.loc[(df['test_round']== xChebNet2MLP_outer_ext.iloc[i]['config/train_loop_config/dataset_test_round']) & (df['es_fold']== xChebNet2MLP_outer_ext.iloc[i]['config/train_loop_config/es_folds']), 'xChebNet2MLP'] = xChebNet2MLP_outer_ext.iloc[i]['test_mcc']

```

```python
# Mean over CV folds for ES (not independent!)
df_fold = df.groupby('test_round').mean()
```

```python
# Experiment 1) MLP vs. GNNs

#full data
MLP_vs_GCN = scipy.stats.wilcoxon(df_fold['MLP'].to_numpy()-df_fold['GCN'].to_numpy())
MLP_vs_GAT = scipy.stats.wilcoxon(df_fold['MLP'].to_numpy()-df_fold['GAT'].to_numpy())
MLP_vs_ChebNet = scipy.stats.wilcoxon(df_fold['MLP'].to_numpy()-df_fold['ChebNet'].to_numpy())
MLP_vs_GCN2MLP = scipy.stats.wilcoxon(df_fold['MLP'].to_numpy()-df_fold['GCN2MLP'].to_numpy())
MLP_vs_GAT2MLP = scipy.stats.wilcoxon(df_fold['MLP'].to_numpy()-df_fold['GAT2MLP'].to_numpy())
MLP_vs_ChebNet2MLP = scipy.stats.wilcoxon(df_fold['MLP'].to_numpy()-df_fold['ChebNet2MLP'].to_numpy())


print('MLP vs. GCN: ', MLP_vs_GCN.pvalue)
print('MLP vs. GAT: ', MLP_vs_GAT.pvalue)
print('MLP vs. ChebNet: ', MLP_vs_ChebNet.pvalue)
print('MLP vs. GCN2MLP: ', MLP_vs_GCN2MLP.pvalue)
print('MLP vs. GAT2MLP: ', MLP_vs_GAT2MLP.pvalue)
print('MLP vs. ChebNet2MLP: ', MLP_vs_ChebNet2MLP.pvalue)

print('\nMLP and GCN2MLP same?', df_fold['MLP'].equals(df_fold['GCN2MLP']))
print('MLP and ChebNet2MLP same?', df_fold['MLP'].equals(df_fold['ChebNet2MLP']), '\n')

#methylation data
mMLP_vs_mGCN = scipy.stats.wilcoxon(df_fold['mMLP'].to_numpy()-df_fold['mGCN'].to_numpy())
mMLP_vs_mGAT = scipy.stats.wilcoxon(df_fold['mMLP'].to_numpy()-df_fold['mGAT'].to_numpy())
mMLP_vs_mChebNet = scipy.stats.wilcoxon(df_fold['mMLP'].to_numpy()-df_fold['mChebNet'].to_numpy())
mMLP_vs_mGCN2MLP = scipy.stats.wilcoxon(df_fold['mMLP'].to_numpy()-df_fold['mGCN2MLP'].to_numpy())
mMLP_vs_mGAT2MLP = scipy.stats.wilcoxon(df_fold['mMLP'].to_numpy()-df_fold['mGAT2MLP'].to_numpy())
mMLP_vs_mChebNet2MLP = scipy.stats.wilcoxon(df_fold['mMLP'].to_numpy()-df_fold['mChebNet2MLP'].to_numpy())


print('\nmMLP vs. mGCN: ', mMLP_vs_mGCN.pvalue)
print('mMLP vs. mGAT: ', mMLP_vs_mGAT.pvalue)
print('mMLP vs. mChebNet: ', mMLP_vs_mChebNet.pvalue)
print('mMLP vs. mGCN2MLP: ', mMLP_vs_mGCN2MLP.pvalue)
print('mMLP vs. mGAT2MLP: ', mMLP_vs_mGAT2MLP.pvalue)
print('mMLP vs. mChebNet2MLP: ', mMLP_vs_mChebNet2MLP.pvalue)

#print('\nmMLP and mChebNet2MLP same?', df_fold['mMLP'].equals(df_fold['mChebNet2MLP']), '\n')

#expression data
xMLP_vs_xGCN = scipy.stats.wilcoxon(df_fold['xMLP'].to_numpy()-df_fold['xGCN'].to_numpy())
xMLP_vs_xGAT = scipy.stats.wilcoxon(df_fold['xMLP'].to_numpy()-df_fold['xGAT'].to_numpy())
xMLP_vs_xChebNet = scipy.stats.wilcoxon(df_fold['xMLP'].to_numpy()-df_fold['xChebNet'].to_numpy())
xMLP_vs_xGCN2MLP = scipy.stats.wilcoxon(df_fold['xMLP'].to_numpy()-df_fold['xGCN2MLP'].to_numpy())
xMLP_vs_xGAT2MLP = scipy.stats.wilcoxon(df_fold['xMLP'].to_numpy()-df_fold['xGAT2MLP'].to_numpy())
xMLP_vs_xChebNet2MLP = scipy.stats.wilcoxon(df_fold['xMLP'].to_numpy()-df_fold['xChebNet2MLP'].to_numpy())


print('\nxMLP vs. xGCN: ', xMLP_vs_xGCN.pvalue)
print('xMLP vs. xGAT: ', xMLP_vs_xGAT.pvalue)
print('xMLP vs. xChebNet: ', xMLP_vs_xChebNet.pvalue)
print('xMLP vs. xGCN2MLP: ', xMLP_vs_xGCN2MLP.pvalue)
print('xMLP vs. xGAT2MLP: ', xMLP_vs_xGAT2MLP.pvalue)
print('xMLP vs. xChebNet2MLP: ', xMLP_vs_xChebNet2MLP.pvalue)

print('\np-value:', 0.05)
print('p-value with Bonferroni correction (within experiment):', 0.05/21)

```

```python
# Experiment 2) full vs. methyl vs. expr (for MLP and GNNs)

MLP_vs_mMLP = scipy.stats.wilcoxon(df_fold['MLP'].to_numpy()-df_fold['mMLP'].to_numpy())
MLP_vs_xMLP = scipy.stats.wilcoxon(df_fold['MLP'].to_numpy()-df_fold['xMLP'].to_numpy())
mMLP_vs_xMLP = scipy.stats.wilcoxon(df_fold['mMLP'].to_numpy()-df_fold['xMLP'].to_numpy())

GNNs_vs_mGNNs = scipy.stats.wilcoxon(pd.concat([df_fold['GCN'], df_fold['GAT'], df_fold['ChebNet'],df_fold['GCN2MLP'], df_fold['GAT2MLP'],df_fold['ChebNet2MLP']]).to_numpy()-pd.concat([df_fold['mGCN'], df_fold['mGAT'], df_fold['mChebNet'],df_fold['mGCN2MLP'], df_fold['mGAT2MLP'],df_fold['mChebNet2MLP']]))
GNNs_vs_xGNNs = scipy.stats.wilcoxon(pd.concat([df_fold['GCN'], df_fold['GAT'], df_fold['ChebNet'],df_fold['GCN2MLP'], df_fold['GAT2MLP'],df_fold['ChebNet2MLP']]).to_numpy()-pd.concat([df_fold['xGCN'], df_fold['xGAT'], df_fold['xChebNet'],df_fold['xGCN2MLP'], df_fold['xGAT2MLP'],df_fold['xChebNet2MLP']]))
mGNNs_vs_xGNNs = scipy.stats.wilcoxon(pd.concat([df_fold['xGCN'], df_fold['xGAT'], df_fold['xChebNet'],df_fold['xGCN2MLP'], df_fold['xGAT2MLP'],df_fold['xChebNet2MLP']]).to_numpy()-pd.concat([df_fold['mGCN'], df_fold['mGAT'], df_fold['mChebNet'],df_fold['mGCN2MLP'], df_fold['mGAT2MLP'],df_fold['mChebNet2MLP']]))


print('MLP vs. mMLP: ', MLP_vs_mMLP.pvalue)
print('MLP vs. xMLP: ', MLP_vs_xMLP.pvalue)
print('mMLP vs. xMLP: ', mMLP_vs_xMLP.pvalue)

print('\nMLP and mMLP same?', df_fold['MLP'].equals(df_fold['mMLP']), '\n')

print('GNNs vs. mGNNs: ', GNNs_vs_mGNNs.pvalue)
print('GNNs vs. xGNNs: ', GNNs_vs_xGNNs.pvalue )
print('mGNNs vs. xGNNs: ', mGNNs_vs_xGNNs.pvalue)


print('\np-value:', 0.05)
print('p-value with Bonferroni correction (within experiment):', 0.05/6)
```

```python
# Experiment 3a) GCN vs. GAT vs. ChebNet (over all data modalities)
GCN_vs_GAT_all_data = scipy.stats.wilcoxon(pd.concat([df_fold['GCN'], df_fold['mGCN'], df_fold['xGCN']]).to_numpy()-pd.concat([df_fold['GAT'], df_fold['mGAT'], df_fold['xGAT']]).to_numpy())
GCN_vs_ChebNet_all_data = scipy.stats.wilcoxon(pd.concat([df_fold['GCN'], df_fold['mGCN'], df_fold['xGCN']]).to_numpy()-pd.concat([df_fold['ChebNet'], df_fold['mChebNet'], df_fold['xChebNet']]).to_numpy())
GAT_vs_ChebNet_all_data = scipy.stats.wilcoxon(pd.concat([df_fold['GAT'], df_fold['mGAT'], df_fold['xGAT']]).to_numpy()-pd.concat([df_fold['ChebNet'], df_fold['mChebNet'], df_fold['xChebNet']]).to_numpy())



print('GCN vs. GAT (all data): ', GCN_vs_GAT_all_data.pvalue)
print('GCN vs. ChebNet (all data): ', GCN_vs_ChebNet_all_data.pvalue)
print('GAT vs. ChebNet (all data): ', GAT_vs_ChebNet_all_data.pvalue)

print('\np-value:', 0.05)
print('p-value with Bonferroni correction (within experiment):', 0.05/3)
```

```python
# Experiment 3b) readout functions (over all data modalities)
GCN_vs_GCN2MLP_all_data = scipy.stats.wilcoxon(pd.concat([df_fold['GCN'], df_fold['mGCN'], df_fold['xGCN']]).to_numpy()-pd.concat([df_fold['GCN2MLP'], df_fold['mGCN2MLP'], df_fold['xGCN2MLP']]).to_numpy())
GAT_vs_GAT2MLP_all_data = scipy.stats.wilcoxon(pd.concat([df_fold['GAT'], df_fold['mGAT'], df_fold['xGAT']]).to_numpy()-pd.concat([df_fold['GAT2MLP'], df_fold['mGAT2MLP'], df_fold['xGAT2MLP']]).to_numpy())
ChebNet_vs_ChebNet2MLP_all_data = scipy.stats.wilcoxon(pd.concat([df_fold['ChebNet'], df_fold['mChebNet'], df_fold['xChebNet']]).to_numpy()-pd.concat([df_fold['ChebNet2MLP'], df_fold['mChebNet2MLP'], df_fold['xChebNet2MLP']]).to_numpy())


print('GCN vs. GCN2MLP (all data): ', GCN_vs_GCN2MLP_all_data.pvalue)
print('GAT vs. GAT2MLP (all data): ', GAT_vs_GAT2MLP_all_data.pvalue)
print('ChebNet vs. ChebNet2MLP (all data): ', ChebNet_vs_ChebNet2MLP_all_data.pvalue)

print('\np-value:', 0.05)
print('p-value with Bonferroni correction (within experiment):', 0.05/3)
```
